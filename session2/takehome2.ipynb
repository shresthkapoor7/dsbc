```python
import numpy as np
```

```python
A = np.array([[1, 2, 3], [3, 4, 5]])
B = np.array([[5, 6, 7], [7, 8, 9]])

print(np.vstack((A, B)))
print(np.hstack((A, B)))
```

```python
print(np.intersect1d(A, B))
```

```python
print(A[np.where((A >= 3) & (A <= 10))])
```

```python
website = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
data = np.genfromtxt(website, delimiter=',', dtype='float', usecols=[0, 1, 2, 3])

print(data[(data[:, 2] > 1.5) & (data[:, 0] < 5.0)])
```

```python
import pandas as pd
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/selva86/datasets/master/Cars93_miss.csv')
print(df.head)
```

```python
print(df.loc[::20, ['Manufacturer', 'Model', 'Type']])
```

```python
df['Min.Price'].fillna(df['Min.Price'].mean(), inplace=True)
df['Max.Price'].fillna(df['Max.Price'].mean(), inplace=True)


print(df[['Min.Price', 'Max.Price']].head())
```
